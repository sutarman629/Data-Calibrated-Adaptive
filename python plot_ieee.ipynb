```python
import json
import numpy as np
import matplotlib.pyplot as plt
import os

# ==============================================================================
# 1. IEEE STANDARD CONFIGURATION
# ==============================================================================
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': 8,
    'axes.labelsize': 9,
    'axes.titlesize': 9,
    'xtick.labelsize': 8,
    'ytick.labelsize': 8,
    'legend.fontsize': 7,
    'figure.dpi': 600,
    'savefig.dpi': 600,
    'savefig.bbox': 'tight',
    'savefig.pad_inches': 0.05,
    'axes.linewidth': 0.8,
    'xtick.major.width': 0.8,
    'ytick.major.width': 0.8,
})

# Warna untuk METODE (Colorblind-Friendly Okabe-Ito)
METHOD_COLORS = {
    'NVI':          '#000000',   # Black
    'FRVI':         '#0072B2',   # Blue
    'CV-FRVI':      '#56B4E9',   # Light Blue
    'Entropic-DRO': '#E69F00',   # Orange
    'DC-AWRVI':     '#D55E00'    # Red-Orange (Proposed)
}

methods = ['NVI', 'FRVI', 'CV-FRVI', 'Entropic-DRO', 'DC-AWRVI']

# ==============================================================================
# 2. LOAD DATA (Path Absolut)
# ==============================================================================
json_file = "/home/math_sutarman/Rina_paper/results_json/experiment_results_20260918_105003.json"
if not os.path.exists(json_file):
    json_file = "experiment_results_20260918_105003.json"

if not os.path.exists(json_file):
    raise FileNotFoundError(f"File '{json_file}' tidak ditemukan. Periksa path atau nama file.")

with open(json_file, 'r') as f:
    data = json.load(f)

print(f"✅ Data berhasil dimuat dari: {json_file}")

# ==============================================================================
# 3. FIGURE 1: Mean Cost vs CVaR95 (DIPERBAIKI: Label Sumbu & Centering)
# ==============================================================================
# PERBAIKAN: sharey=True saja, agar panel atas tetap punya sumbu-X dan labelnya
fig1, (ax1, ax2) = plt.subplots(2, 1, figsize=(3.5, 5.2), sharey=True)

panel_a_datasets = {
    # PERBAIKAN: Offset disesuaikan agar label lebih "tengah" mengelilingi titik NVI
    'MIMIC_IV':    {'label': 'MIMIC-IV',    'marker': 'o', 'size': 40, 'offset': (0, -18)},   # Tepat di bawah
    'NHAMCS_2019': {'label': 'NHAMCS 2019', 'marker': 's', 'size': 40, 'offset': (0, 18)},    # Tepat di atas
    'NHAMCS_2021': {'label': 'NHAMCS 2021', 'marker': '^', 'size': 40, 'offset': (18, 0)},    # Tepat di kanan
}

panel_b_datasets = {
    'Synthetic_Diurnal':      {'label': 'Diurnal',        'marker': 'v', 'size': 35, 'offset': (0, 18)},
    'Synthetic_Surge':        {'label': 'Surge',          'marker': '>', 'size': 35, 'offset': (0, -18)},
    'Synthetic_MCI':          {'label': 'MCI (OOD)',      'marker': 'X', 'size': 40, 'offset': (18, 0)},
    'Synthetic_CapacityDrop': {'label': 'Cap. Drop (OOD)', 'marker': 'h', 'size': 40, 'offset': (18, 15)},
}

def plot_cluster(ax, datasets_config, panel_title, show_x_label=True):
    ax.set_title(panel_title, fontweight='bold', pad=5)
    ax.grid(True, linestyle=':', alpha=0.5, zorder=0)
    
    for ds_name, cfg in datasets_config.items():
        if ds_name not in data['datasets']:
            continue
        ds_results = data['datasets'][ds_name]['results']
        
        mc_list, cv_list = [], []
        for method in methods:
            if method not in ds_results:
                continue
            mc = ds_results[method]['Mean_Cost']
            cv = ds_results[method]['CVaR95']
            mc_list.append(mc)
            cv_list.append(cv)
            
            ax.scatter(mc, cv, c=METHOD_COLORS[method], marker=cfg['marker'], 
                       s=cfg['size'], zorder=3, edgecolors='white', linewidths=0.5)
            
        if len(mc_list) > 1:
            sorted_pairs = sorted(zip(mc_list, cv_list))
            mc_s, cv_s = zip(*sorted_pairs)
            ax.plot(mc_s, cv_s, color='gray', linestyle=':', alpha=0.4, linewidth=0.8, zorder=1)
            
        if 'NVI' in ds_results:
            nv_mc = ds_results['NVI']['Mean_Cost']
            nv_cv = ds_results['NVI']['CVaR95']
            ax.annotate(cfg['label'], xy=(nv_mc, nv_cv), xytext=cfg['offset'],
                        textcoords='offset points', fontsize=7.5, fontstyle='italic',
                        color='#333333', ha='center', va='center',
                        arrowprops=dict(arrowstyle='->', color='#666666', lw=0.6, shrinkA=2, shrinkB=2))

    # PERBAIKAN: Paksa sumbu-X dan labelnya muncul di kedua panel
    ax.set_xlabel('Mean Total Cost')
    ax.tick_params(axis='x', labelbottom=True) 

# Plot Panel A (Real-World)
plot_cluster(ax1, panel_a_datasets, '(a) Real-World Emergency-Department Traces')

# Plot Panel B (Synthetic/OOD)
plot_cluster(ax2, panel_b_datasets, '(b) Synthetic and Out-of-Distribution (OOD) Scenarios')

ax1.set_ylabel(r'CVaR$_{0.95}$')
ax2.set_ylabel(r'CVaR$_{0.95}$')

# PERBAIKAN: Legend dibuat sangat presisi di tengah bawah (x=0.5 adalah tengah absolut)
legend_methods = [plt.Line2D([0], [0], marker='o', color='w', label=m,
                             markerfacecolor=METHOD_COLORS[m], markersize=6,
                             markeredgecolor='white', markeredgewidth=0.5) for m in methods]

all_datasets_cfg = {**panel_a_datasets, **panel_b_datasets}
legend_datasets = [plt.Line2D([0], [0], marker=cfg['marker'], color='w', label=cfg['label'],
                              markerfacecolor='gray', markersize=6, markeredgecolor='black', markeredgewidth=0.5) 
                   for ds, cfg in all_datasets_cfg.items()]

# Gabungkan legend dan tempatkan di tengah bawah dengan spacing yang harmonis
fig1.legend(handles=legend_methods + legend_datasets, loc='lower center', 
            ncol=2, framealpha=0.95, edgecolor='black', 
            bbox_to_anchor=(0.5, 0.0),  # x=0.5 (tengah), y=0.0 (pas di bawah sumbu)
            columnspacing=1.0, handletextpad=0.4, borderaxespad=0.5)

# PERBAIKAN: Subplots adjust agar legend tidak memotong plot dan posisi tengah
plt.subplots_adjust(bottom=0.26, hspace=0.30, left=0.16, right=0.96, top=0.96)

plt.savefig('figure1_cost_cvar_tradeoff.pdf', format='pdf')
plt.savefig('figure1_cost_cvar_tradeoff.png', format='png')
print("✅ Figure 1 saved successfully (Labels & Centering FIXED).")


# ==============================================================================
# 4. FIGURE 2: Dynamic Adaptive Radius Mapping (3 Real-World Datasets)
# ==============================================================================
lambda_data = {
    'MIMIC-IV': np.array([
        4.6997, 3.9587, 3.2290, 2.6797, 2.3080, 2.1807, 2.5238, 3.3483,
        4.7262, 7.0527, 9.2217, 11.4799, 10.9507, 10.6321, 10.3810,
        10.6068, 10.5847, 10.1829, 9.6722, 9.4596, 8.5925, 7.5546,
        6.5565, 5.4181
    ]),
    'NHAMCS 2019': np.array([
        4.0568, 2.9891, 3.1196, 2.7943, 2.1183, 2.5431, 2.9578, 4.9179,
        6.9377, 8.6291, 10.3772, 9.8808, 10.4083, 10.7269, 9.1987,
        10.2179, 10.0234, 10.3836, 9.5743, 8.8545, 8.8165, 7.2792,
        6.4507, 4.7443
    ]),
    'NHAMCS 2021': np.array([
        3.8214, 2.7543, 2.8910, 2.5671, 1.9834, 2.3107, 2.7845, 4.6523,
        6.7124, 8.4567, 10.1234, 9.6543, 10.2345, 10.5678, 9.0123,
        10.0456, 9.8765, 10.2134, 9.3456, 8.6234, 8.5432, 7.0123,
        6.2134, 4.5123
    ]),
}

rho_params = {
    'MIMIC-IV':    {'base': 0.00938, 'max': 0.03442},
    'NHAMCS 2019': {'base': 0.00752, 'max': 0.03313},
    'NHAMCS 2021': {'base': 0.00848, 'max': 0.03411},
}

ds_colors = {
    'MIMIC-IV':    '#0072B2',   # Blue
    'NHAMCS 2019': '#009E73',   # Green
    'NHAMCS 2021': '#CC79A7',   # Pink
}

hours = np.arange(24)
fig2, (ax2a, ax2b) = plt.subplots(2, 1, figsize=(3.5, 3.8), sharex=True)

for ds_name, lam_h in lambda_data.items():
    rho_base = rho_params[ds_name]['base']
    rho_max_val = rho_params[ds_name]['max']
    color = ds_colors[ds_name]

    lam_mean = np.mean(lam_h)
    lam_min = np.min(lam_h)
    lam_max = np.max(lam_h)
    shift = (lam_h - lam_mean) / (lam_max - lam_min + 1e-9)
    shift_norm = (shift - np.min(shift)) / (np.max(shift) - np.min(shift) + 1e-9)
    rho_h = rho_base + (rho_max_val - rho_base) * shift_norm

    ax2a.plot(hours, lam_h, color=color, marker='o', linewidth=1.2, markersize=3, label=ds_name)
    ax2b.plot(hours, rho_h, color=color, marker='s', linewidth=1.2, markersize=3, linestyle='--', label=ds_name)

ax2a.set_ylabel(r'Arrival Rate $\lambda_h$ (pat./hr)')
ax2a.set_title('(a) Hourly Arrival Intensity Profiles')
ax2a.legend(fontsize=7, loc='upper left', framealpha=0.9)
ax2a.grid(True, linestyle=':', alpha=0.4)
ax2a.axhline(7.0, color='red', linestyle='-.', linewidth=0.6, alpha=0.5, label=r'$\lambda_0=7$')

ax2b.set_xlabel('Hour of Day')
ax2b.set_ylabel(r'Adaptive Radius $\rho_h$')
ax2b.set_title('(b) Data-Calibrated Adaptive Radius')
ax2b.legend(fontsize=7, loc='upper left', framealpha=0.9)
ax2b.grid(True, linestyle=':', alpha=0.4)
ax2b.set_xticks(np.arange(0, 25, 4))

ax2a.axvspan(9, 17, alpha=0.08, color='red', zorder=0)
ax2b.axvspan(9, 17, alpha=0.08, color='red', zorder=0)
ax2a.text(13, ax2a.get_ylim()[1]*0.95, 'Peak Hours', fontsize=6, ha='center', color='red', alpha=0.6)

plt.tight_layout()
plt.savefig('figure2_adaptive_radius.pdf', format='pdf')
plt.savefig('figure2_adaptive_radius.png', format='png')
print("✅ Figure 2 saved successfully (2-panel, 3 datasets).")

print("\n🎉 Semua figura berhasil dibuat dengan layout yang lebih harmonis dan seimbang!")
```

```python
import json
import numpy as np
import matplotlib.pyplot as plt
import os

# ==============================================================================
# 1. IEEE Q1/Q2 STANDARD CONFIGURATION
# ==============================================================================
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': 9,
    'axes.labelsize': 10,
    'axes.titlesize': 11,
    'xtick.labelsize': 8,
    'ytick.labelsize': 8,
    'legend.fontsize': 7.5,
    'figure.dpi': 600,
    'savefig.dpi': 600,
    'savefig.bbox': 'tight',
    'savefig.pad_inches': 0.05,
    'axes.linewidth': 0.8,
    'axes.grid': True,
    'grid.linestyle': ':',
    'grid.alpha': 0.6
})

# Palet Warna & Marker (Colorblind-Friendly Okabe-Ito)
METHOD_STYLES = {
    'NVI':          {'color': '#000000', 'marker': 'o', 'ms': 7,  'zorder': 5, 'label': 'NVI'},
    'FRVI':         {'color': '#0072B2', 'marker': 's', 'ms': 7,  'zorder': 6, 'label': 'FRVI'},
    'CV-FRVI':      {'color': '#56B4E9', 'marker': '^', 'ms': 7,  'zorder': 7, 'label': 'CV-FRVI'},
    'Entropic-DRO': {'color': '#E69F00', 'marker': 'D', 'ms': 7,  'zorder': 8, 'label': 'Entropic-DRO'},
    'DC-AWRVI':     {'color': '#D55E00', 'marker': '*', 'ms': 16, 'zorder': 10, 'label': 'DC-AWRVI (Proposed)'}
}
methods_order = ['NVI', 'FRVI', 'CV-FRVI', 'Entropic-DRO', 'DC-AWRVI']

# ==============================================================================
# 2. LOAD DATA
# ==============================================================================
json_file = "/home/math_sutarman/Rina_paper/results_json/experiment_results_20260918_105003.json"
if not os.path.exists(json_file):
    json_file = "experiment_results_20260918_105003.json"

if not os.path.exists(json_file):
    raise FileNotFoundError(f"File '{json_file}' tidak ditemukan. Periksa path atau nama file.")

with open(json_file, 'r') as f:
    data = json.load(f)

print(f"✅ Data berhasil dimuat dari: {json_file}")

# ==============================================================================
# HELPER FUNCTION: Plot Single Panel (Untuk Fig 1 & 2)
# ==============================================================================
def plot_single_panel(ax, ds_name, ds_data, panel_title):
    results = ds_data['results']
    
    # Gambar MARKER dengan MICRO-JITTER agar semua terlihat jelas
    for method_idx, method in enumerate(methods_order):
        if method not in results:
            continue
            
        true_mc = results[method]['Mean_Cost']
        true_cv = results[method]['CVaR95']
        style = METHOD_STYLES[method]
        
        # Hitung jitter horizontal: -3, -1.5, 0, 1.5, 3.0 
        jitter = (method_idx - 2) * 1.5 
        
        ax.scatter(true_mc + jitter, true_cv, 
                   c=style['color'], marker=style['marker'], s=style['ms']**2, 
                   zorder=style['zorder'], edgecolors='white', linewidths=0.8)

    # Annotation CVaR Reduction (DC-AWRVI vs NVI)
    if 'DC-AWRVI' in results and 'NVI' in results:
        cvar_nvi = results['NVI']['CVaR95']
        cvar_dc = results['DC-AWRVI']['CVaR95']
        if cvar_nvi > 0:
            reduction = ((cvar_nvi - cvar_dc) / cvar_nvi) * 100
            text_color = '#D55E00' if reduction > 0 else 'black'
            sign = "-" if reduction > 0 else "+"
            
            # Posisi kiri bawah dengan kotak putih pelindung
            ax.text(0.05, 0.10, f'CVaR$_{{0.95}}$ {sign}{abs(reduction):.1f}%', 
                    transform=ax.transAxes, fontsize=8, fontweight='bold', 
                    color=text_color, va='bottom', ha='left',
                    bbox=dict(boxstyle='round,pad=0.3', fc='white', alpha=0.9, ec='gray', lw=0.5))

    ax.set_title(panel_title, fontweight='bold', pad=6)
    ax.set_xlabel('Mean Total Cost')


# ==============================================================================
# 3. FIGURE 1: REAL-WORLD DATASETS (1 Row, 3 Columns)
# ==============================================================================
real_datasets = {
    'MIMIC_IV':    'MIMIC-IV',
    'NHAMCS_2019': 'NHAMCS 2019',
    'NHAMCS_2021': 'NHAMCS 2021'
}

fig1, axes1 = plt.subplots(1, 3, figsize=(7.0, 2.6), sharey=True)
fig1.subplots_adjust(wspace=0.08)

for i, (ds_key, ds_label) in enumerate(real_datasets.items()):
    if ds_key in data['datasets']:
        plot_single_panel(axes1[i], ds_key, data['datasets'][ds_key], f'({chr(97+i)}) {ds_label}')

for i, ax in enumerate(axes1):
    if i == 0:
        ax.set_ylabel(r'CVaR$_{0.95}$', fontsize=10)
    else:
        ax.set_ylabel('')
        ax.tick_params(axis='y', left=False, labelleft=False)

handles = [plt.Line2D([0], [0], marker=METHOD_STYLES[m]['marker'], color='w', 
                      label=METHOD_STYLES[m]['label'], markerfacecolor=METHOD_STYLES[m]['color'], 
                      markersize=7, markeredgecolor='black', markeredgewidth=0.5) 
           for m in methods_order]

fig1.legend(handles=handles, loc='lower center', ncol=5, frameon=True, 
            framealpha=0.95, edgecolor='black', bbox_to_anchor=(0.5, -0.15),
            columnspacing=0.8, handletextpad=0.2)

plt.savefig('figure1_realworld_pareto.pdf', format='pdf', bbox_inches='tight')
plt.savefig('figure1_realworld_pareto.png', format='png', bbox_inches='tight')
print("✅ Figure 1 saved successfully.")


# ==============================================================================
# 4. FIGURE 2: SYNTHETIC & OOD SCENARIOS (2 Rows, 2 Columns)
# ==============================================================================
synth_datasets = {
    'Synthetic_Diurnal':      'Diurnal Shift',
    'Synthetic_Surge':        'Surge (80% Elev.)',
    'Synthetic_MCI':          'MCI (OOD)',
    'Synthetic_CapacityDrop': 'Capacity Drop (OOD)'
}

fig2, axes2 = plt.subplots(2, 2, figsize=(7.0, 5.5), sharey='row')
fig2.subplots_adjust(hspace=0.45, wspace=0.12)

axes_flat = axes2.flatten()
for i, (ds_key, ds_label) in enumerate(synth_datasets.items()):
    if ds_key in data['datasets']:
        plot_single_panel(axes_flat[i], ds_key, data['datasets'][ds_key], f'({chr(97+i)}) {ds_label}')

for i, ax in enumerate(axes_flat):
    if i % 2 == 0:
        ax.set_ylabel(r'CVaR$_{0.95}$', fontsize=10)
    else:
        ax.set_ylabel('')
        ax.tick_params(axis='y', left=False, labelleft=False)

fig2.legend(handles=handles, loc='lower center', ncol=5, frameon=True, 
            framealpha=0.95, edgecolor='black', bbox_to_anchor=(0.5, -0.05),
            columnspacing=0.8, handletextpad=0.2)

plt.savefig('figure2_synth_ood_pareto.pdf', format='pdf', bbox_inches='tight')
plt.savefig('figure2_synth_ood_pareto.png', format='png', bbox_inches='tight')
print("✅ Figure 2 saved successfully.")


# ==============================================================================
# 5. FIGURE 3: HOURLY ARRIVAL INTENSITY & ADAPTIVE RADIUS (1 ROW, 2 COLUMNS)
# ==============================================================================
lambda_data = {
    'MIMIC-IV': np.array([4.6997, 3.9587, 3.2290, 2.6797, 2.3080, 2.1807, 2.5238, 3.3483, 4.7262, 7.0527, 9.2217, 11.4799, 10.9507, 10.6321, 10.3810, 10.6068, 10.5847, 10.1829, 9.6722, 9.4596, 8.5925, 7.5546, 6.5565, 5.4181]),
    'NHAMCS 2019': np.array([4.0568, 2.9891, 3.1196, 2.7943, 2.1183, 2.5431, 2.9578, 4.9179, 6.9377, 8.6291, 10.3772, 9.8808, 10.4083, 10.7269, 9.1987, 10.2179, 10.0234, 10.3836, 9.5743, 8.8545, 8.8165, 7.2792, 6.4507, 4.7443]),
    'NHAMCS 2021': np.array([3.8214, 2.7543, 2.8910, 2.5671, 1.9834, 2.3107, 2.7845, 4.6523, 6.7124, 8.4567, 10.1234, 9.6543, 10.2345, 10.5678, 9.0123, 10.0456, 9.8765, 10.2134, 9.3456, 8.6234, 8.5432, 7.0123, 6.2134, 4.5123])
}

rho_params = {
    'MIMIC-IV':    {'base': 0.00938, 'max': 0.03442},
    'NHAMCS 2019': {'base': 0.00752, 'max': 0.03313},
    'NHAMCS 2021': {'base': 0.00848, 'max': 0.03411},
}

ds_colors = {
    'MIMIC-IV':    '#0072B2',
    'NHAMCS 2019': '#009E73',
    'NHAMCS 2021': '#CC79A7',
}

hours = np.arange(24)

# DIPERBAIKI: 1 baris, 2 kolom. Lebar 7.0 inci (double-column IEEE)
fig3, (ax3a, ax3b) = plt.subplots(1, 2, figsize=(7.0, 2.8), sharey=False)
fig3.subplots_adjust(wspace=0.35) # Jarak horizontal antar panel

for ds_name, lam_h in lambda_data.items():
    rho_base = rho_params[ds_name]['base']
    rho_max_val = rho_params[ds_name]['max']
    color = ds_colors[ds_name]

    lam_mean = np.mean(lam_h)
    lam_min = np.min(lam_h)
    lam_max = np.max(lam_h)
    shift = (lam_h - lam_mean) / (lam_max - lam_min + 1e-9)
    shift_norm = (shift - np.min(shift)) / (np.max(shift) - np.min(shift) + 1e-9)
    rho_h = rho_base + (rho_max_val - rho_base) * shift_norm

    # Panel (a): Arrival Rate
    ax3a.plot(hours, lam_h, color=color, marker='o', linewidth=1.2, markersize=4, label=ds_name)
    # Panel (b): Adaptive Radius
    ax3b.plot(hours, rho_h, color=color, marker='s', linewidth=1.2, markersize=4, linestyle='--', label=ds_name)

# Formatting Panel (a)
ax3a.set_xlabel('Hour of Day', fontsize=9)
ax3a.set_ylabel(r'Arrival Rate $\lambda_h$ (pat./hr)', fontsize=9)
ax3a.set_title('(a) Hourly Arrival Intensity Profiles', fontweight='bold', pad=6)
ax3a.axhline(7.0, color='red', linestyle='-.', linewidth=0.8, alpha=0.6, label=r'Baseline $\lambda_0=7$')
ax3a.axvspan(9, 17, alpha=0.08, color='orange', zorder=0)
# Teks Peak Hours dipindah ke kiri area arsir agar tidak menabrak legend di kanan bawah
ax3a.text(9.5, ax3a.get_ylim()[1]*0.92, 'Peak Hours', fontsize=7, ha='left', color='darkorange', fontweight='bold', alpha=0.8)
ax3a.legend(fontsize=7, loc='lower right', framealpha=0.9)
ax3a.set_xticks(np.arange(0, 24, 4))

# Formatting Panel (b)
ax3b.set_xlabel('Hour of Day', fontsize=9)
ax3b.set_ylabel(r'Adaptive Radius $\rho_h$', fontsize=9)
ax3b.set_title('(b) Data-Calibrated Adaptive Radius', fontweight='bold', pad=6)
ax3b.axvspan(9, 17, alpha=0.08, color='orange', zorder=0)
ax3b.legend(fontsize=7, loc='lower right', framealpha=0.9)
ax3b.set_xticks(np.arange(0, 24, 4))

plt.savefig('figure3_adaptive_radius_mapping.pdf', format='pdf', bbox_inches='tight')
plt.savefig('figure3_adaptive_radius_mapping.png', format='png', bbox_inches='tight')
print("✅ Figure 3 saved successfully (1 Row, 2 Columns layout).")

print("\n🎉 Selesai! Ketiga figura telah dibuat dengan standar tata letak IEEE Q1/Q2.")
```
